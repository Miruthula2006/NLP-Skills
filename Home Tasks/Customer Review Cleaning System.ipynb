```python
# Customer Review Cleaning System
# IMDb 50K Movie Review Dataset

import pandas as pd
import nltk
import re

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Step 1: Load the IMDb Dataset
data = pd.read_csv("IMDB_Dataset_CLEANED.csv")

print("First 5 rows:")
print(data.head())

# Step 2: Inspect Missing Values and Duplicates
print("\nMissing Values:")
print(data.isnull().sum())

print("\nDuplicate Reviews:")
print(data["review"].duplicated().sum())


# Remove duplicate reviews
data = data.drop_duplicates(subset=["review"])

# Remove rows with missing reviews
data = data.dropna(subset=["review"])

# Download NLTK Resources
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')

# Create Stopwords and Lemmatizer
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

# Steps 3 - 7: Preprocessing
def preprocess(text):
    # Step 3: Convert text to lowercase
    text = text.lower()
    # Step 4: Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)
    # Step 4: Remove URLs
    text = re.sub(r'http\S+|www\S+', ' ', text)
    # Step 4: Remove punctuation, numbers and special characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    # Step 6: Tokenization
    words = word_tokenize(text)
    # Step 5: Remove stopwords
    words = [
        word
        for word in words
        if word not in stop_words
    ]
    # Step 7: Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]
    # Return cleaned review
    return " ".join(words)

# Step 8: Store Cleaned Reviews
data["cleaned_review"] = data["review"].apply(preprocess)

# Step 9: Compare Original and Cleaned Reviews
print("\nOriginal and Cleaned Reviews:")
print("--------------------------------")

print(
    data[
        ["review", "sentiment", "cleaned_review"]
    ].head(10)
)

# Step 10: Export Preprocessed Dataset
data.to_csv(
    "IMDB_Dataset_Preprocessed.csv",
    index=False
)

print("\nPreprocessed dataset exported successfully!")

print("\nFinal Dataset Columns:")
print(data.columns)
```
